```python
print("=== CALCULADORA DE DERIVADAS PARA POLINÓMIOS ===")
print("Exemplo: Para f(x) = 2x³ + 3x² - 5x + 4")
print("Os coeficientes são: [2, 3, -5, 4]\n")

# 1. Entrada de dados: Coeficientes do polinómio
# O utilizador digita os números separados por espaços (do maior grau para o menor)
entrada = input("Introduza os coeficientes separados por espaço: ")
coefs = [float(c) for c in entrada.split()]

# 2. Função para avaliar o polinómio num ponto x
def avaliar_polinomio(coeficientes, x):
    grau_maximo = len(coeficientes) - 1
    resultado = 0.0
    for i, c in enumerate(coeficientes):
        potencia = grau_maximo - i
        resultado += c * (x ** potencia)
    return resultado

# 3. Função para obter a lista de coeficientes da FUNÇÃO DERIVADA f'(x)
# Aplica a regra (a * x^n)' = a * n * x^(n-1)
def derivar_polinomio(coeficientes):
    grau_maximo = len(coeficientes) - 1
    coefs_derivada = []
    for i, c in enumerate(coeficientes[:-1]):  # O último elemento (constante) dá derivada 0
        potencia = grau_maximo - i
        coefs_derivada.append(c * potencia)
    return coefs_derivada

# 4. Leitura do ponto x0
x0 = float(input("Calcular a derivada no ponto x = "))

# --- APROXIMAÇÃO NUMÉRICA (O método que está no manual) ---
print("\n" + "="*50)
print(f"{'h (Delta x)':<12} | {'Taxa de Variação (Aproximação)':<30}")
print("="*50)

for i in range(1, 9):
    h = 10**(-i)
    # Cálculo da taxa média de variação
    f_xh = avaliar_polinomio(coefs, x0 + h)
    f_x  = avaliar_polinomio(coefs, x0)
    d = (f_xh - f_x) / h
    print(f"10^-{i:<2} ({h:<8.8f}) | {d:.6f}")

print("="*50)

# --- CÁLCULO EXATO DA DERIVADA (Para os alunos confirmarem) ---
coefs_d = derivar_polinomio(coefs)
valor_exato = avaliar_polinomio(coefs_d, x0)

print(f"\nResultado Exato (via Regra dos Polinómios): f'({x0}) = {valor_exato}")
```
