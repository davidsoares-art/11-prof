```python
import sympy as sp
from sympy.parsing.sympy_parser import (
    parse_expr,
    standard_transformations,
    implicit_multiplication_application,
    convert_xor
)

def calcular_derivada():
    print("=== Calculadora de Derivadas (Melhorada) ===")

    # 1. Ler a variável
    var_str = input("Introduza a variável (ex: x, t, y): ").strip()
    var = sp.Symbol(var_str)

    expr_str = input(f"Introduza a função f({var_str}): ").strip()

    # Configurar transformações automáticas:
    # - implicit_multiplication_application: permite escrever '3x' em vez de '3*x'
    # - convert_xor: converte automaticamente '^' em '**'
    transformacoes = standard_transformations + (
        implicit_multiplication_application,
        convert_xor
    )

    try:
        # Converter a string usando as regras avançadas
        f = parse_expr(expr_str, transformations=transformacoes, local_dict={var_str: var})

        # 2. Calcular a derivada
        f_derivada = sp.diff(f, var)

        print("\n" + "="*30)
        print(f"Função original f({var_str}) = {f}")
        print(f"Função derivada f'({var_str}) = {f_derivada}")
        print("="*30)

        # 3. Opção de calcular num ponto específico
        ponto_input = input(f"\nDesejas calcular a derivada num ponto específico? (Insere o número ou prime Enter para saltar): ").strip()

        if ponto_input != "":
            ponto_val = float(ponto_input)
            resultado_ponto = f_derivada.subs(var, ponto_val)

            print(f"\nResultado em {var_str} = {ponto_val}:")
            print(f"  • Valor exato:   f'({ponto_val}) = {resultado_ponto}")
            print(f"  • Valor decimal: f'({ponto_val}) ≈ {float(resultado_ponto):.4f}")

    except Exception as e:
        print(f"\n[Erro] Não foi possível processar a expressão: {e}")

if __name__ == "__main__":
    calcular_derivada()
```
