```python
import math

print("=== CALCULADORA DE TAXA DE VARIAÇÃO INSTANTÂNEA ===")
print("Dica: Use 'x' como variável e '*' para multiplicar (ex: 2*x**3 + 3*x**2)")

# 1. O utilizador escreve qualquer expressão matemática em função de x
expressao = input("\nIntroduza a função f(x): ")

# Definição dinâmica da função f(x) usando eval
def f(x):
    # Permite usar funções matemáticas como math.sin, math.cos, math.exp, etc.
    return eval(expressao, {"x": x, "math": math, "sin": math.sin, "cos": math.cos, "sqrt": math.sqrt})

# 2. Leitura do ponto onde queremos calcular a derivada
x0 = float(input("\nCalcular a taxa de variação no ponto x = "))

print("\n" + "="*50)
print(f"{'h (Delta x)':<15} | {'f(x0 + h)':<12} | {'Taxa de Variação (d)':<20}")
print("="*50)

# 3. Ciclo FOR para aproximar h de zero (10^-1, 10^-2, ..., 10^-8)
for i in range(1, 10):
    h = 10**(-i)

    # Cálculo da taxa média de variação
    d = (f(x0 + h) - f(x0)) / h

    # Apresentação formatada dos resultados
    print(f"10^-{i:<2} ({h:<8.8f}) | {f(x0+h):<12.5f} | {d:.6f}")

print("="*50)
print("Observe como os valores na última coluna convergem para um valor fixo!")
```
